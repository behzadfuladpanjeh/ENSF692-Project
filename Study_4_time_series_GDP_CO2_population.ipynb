#Study 4: Time Series for Specific Countries

How have GDP, CO₂, and population changed over time for countries like Canada, China, or India?

### Objective:

This study visualizes how GDP, CO₂ emissions, and population have changed over time for selected countries. The trends can help us understand development patterns and sustainability progress over the years.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from modules.data_loader import DataLoader
```

## Load Preprocessed Data
We load 4 databases and merge into a single DataFrame with the following columns:
 `Continent`, `Country`, `year`, `co2`, `gdp`, `population`, `total_co2`, `total_gdp`

```python
# Load the preprocessed CSV 
loader = DataLoader()            # Create an instance of the DataLoader class
data = loader.load_and_prepare_data()  # Call the method to load and clean the data
data.head()
```

## Scatter Plot: GDP, CO₂ emissions, and population (2020)
Visualizing the  GDP, CO₂ emissions, and population for selected countries:

```python
# Filter for selected countries and years
countries = ["Canada", "China", "India", "USA", "Germany", "Russia", "Japan" ]
data_filtered = data[data["Country"].isin(countries) & (data["year"] >= 1950)]

# Plot 1: GDP per Capita over Time
plt.figure(figsize=(10, 6))
for country in countries:
    subset = data_filtered[data_filtered["Country"] == country]
    plt.plot(subset["year"], subset["gdp"], label=country)
plt.title("GDP per Capita (1950–2023)")
plt.xlabel("Year")
plt.ylabel("GDP per Capita (USD)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot 2: CO₂ Emissions per Capita over Time
plt.figure(figsize=(10, 6))
for country in countries:
    subset = data_filtered[data_filtered["Country"] == country]
    plt.plot(subset["year"], subset["co2"], label=country)
plt.title("CO₂ Emissions per Capita (1950–2023)")
plt.xlabel("Year")
plt.ylabel("CO₂ per Capita (tons)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot 3: Population over Time
plt.figure(figsize=(10, 6))
for country in countries:
    subset = data_filtered[data_filtered["Country"] == country]
    plt.plot(subset["year"], subset["population"] / 1e6, label=country)  # Convert to millions
plt.title("Population (1950–2023)")
plt.xlabel("Year")
plt.ylabel("Population (Millions)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Key Observations:
**Developed Nations (USA, Canada, Germany, Japan):**

  
GDP per capita remains high and has generally increased over time in the United States, Canada, and Germany; Japan’s economic growth, however, has plateaued in more recent years.
In terms of CO2 emissions per capita, a distinct trend of decoupling economic growth from carbon output is observed. While GDP per capita continued to rise, per capita CO2 emissions have either peaked and declined (notably in the United States, Canada, and Germany—with Germany’s reduction being particularly marked) or stabilized, as seen in Japan. This pattern likely reflects improvements in energy efficiency, transitions toward service-oriented economies, and greater reliance on cleaner energy sources.
Population growth in these countries is relatively modest, especially when contrasted with the rapid population growth occurring in China and India.

**Emerging Economies (China, India):**

Both countries have experienced rapid and significant increases in GDP per capita from comparatively low starting points, with China’s growth being especially pronounced.
CO2 emissions per capita in China and India have risen consistently and substantially. This indicates that their economic development has been closely tied to carbon-intensive activities. While their per capita emissions remain below those of developed nations, the combination of economic growth and large populations results in considerable increases in total emissions.
Population size and growth in both countries are substantial, further extending the impact of rising per capita emissions on global totals.

**Russia:**

Both GDP per capita and CO2 emissions per capita dropped sharply during the early 1990s. Later on, both metrics have recovered, although Russia’s CO2 emissions per capita have not exhibited the consistent decline seen in many Western nations.
The population has remained relatively stable throughout this period, distinguishing Russia from the rapid population growth observed in China and India.
