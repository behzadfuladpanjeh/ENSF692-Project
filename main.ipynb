# Global Trends in CO₂ Emissions, GDP, and Population

## Introduction

Understanding the relationship between environmental impact, economic development, and demographic shifts is vital for addressing global sustainability challenges. This project analyzes comprehensive datasets covering:

- CO₂ emissions per capita  
- GDP per capita  
- Population  
- Continental grouping  

The objective is to identify significant patterns and trends through data cleaning, integration, restructuring, and visual exploration. Using available data, this study investigates how various countries and continents compare and change over time.

Central research questions include:

- To what extent is economic growth associated with increased carbon emissions?  
- Which nations demonstrate high GDP with relatively low emissions?  
- In what ways does population size affect environmental impact?  

## Project Structure

This repository contains 8 Jupyter Notebooks, each dedicated to a specific study:

1.  **Study_1_economic_growth_vs_emissions.ipynb**: Explores the correlation between GDP per capita and CO₂ emissions per capita.
    
2.  **Study_2_high_population_low_emissions.ipynb**: Identifies countries with high populations but low per capita CO₂ emissions.
    
3.  **Study_3_CO2_per_GDP.ipynb**: Analyzes carbon intensity by identifying countries with the highest CO₂ emissions relative to their economic output.
    
4.  **Study_4_time_series_GDP_CO2_population.ipynb**: Visualizes time-series data for GDP, CO₂, and population for selected countries.
    
5.  **Study_5_continental_emissions.ipynb**: Compares average CO₂ per capita and economic development across continents.
    
6.  **Study_6_CO2_population.ipynb**: Investigates patterns between national population size and total CO₂ emissions using logarithmic scales.
    
7.  **Study_7_most_imporved_countries.ipynb**: Identifies countries that have significantly reduced their per capita CO₂ emissions over time.

8.  **Study_8_country_years.ipynb**: CO₂ per capita, and GDP and population for specific country and specific years.

A `modules/data_loader.py` file is used to load and preprocess the data consistently across all studies.
 `data` folder contain 4 databases: CO₂ emissions, GDP, population across various countries and a country-continents map.
