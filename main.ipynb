# Global Trends in CO₂ Emissions, GDP, and Population

## Introduction

Understanding the relationship between environmental impact, economic development, and demographic shifts is vital for addressing global sustainability challenges. This project analyzes comprehensive datasets covering:

- CO₂ emissions per capita  
- GDP per capita  
- Population  
- Continental grouping  

The objective is to identify significant patterns and trends through data cleaning, integration, restructuring, and visual exploration. Using available data, this study investigates how various countries and continents compare and change over time.

Central research questions include:

- To what extent is economic growth associated with increased carbon emissions?  
- Which nations demonstrate high GDP with relatively low emissions?  
- In what ways does population size affect environmental impact?  

These questions, among others, will be examined across nine targeted analytical scenarios:

1. **Economic Growth vs. Emissions**  
   Does higher GDP per capita correlate with higher or lower CO₂ emissions?

2. **High Population, Low Emissions**  
   Which high-population countries maintain low per capita CO₂ emissions?

3. **CO₂ per GDP (Carbon Intensity)**  
   Which countries have the highest CO₂ emissions relative to their economic output?

4. **Time-Series for Specific Countries**  
   How have GDP, CO₂, and population changed over time for countries like Canada, China, or India?

5. **Continental Emissions Comparison**  
   How do continents compare in terms of average CO₂ per capita and economic development?

6. **CO₂ Emissions vs. Population (Log-Scale)**  
   Are there patterns between national population size and emissions when using logarithmic scale?

7. **Most Improved Countries (Emissions Reduction)**  
   Which countries have reduced their per capita CO₂ emissions the most over time?

8. **Developed vs. Developing Countries**  
   Are there observable differences in CO₂ emissions trends between developed and developing countries?

9. **Outliers in Emissions**  
   Which countries emit unusually high or low CO₂ per capita for their GDP?
